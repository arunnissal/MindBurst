# 🚀 MindBurst — 1-Click Android APK Builder
This notebook compiles the **MindBurst** mobile app into a ready-to-install Android `.apk` file using Buildozer on Google Colab (free Ubuntu cloud runner).

### Instructions:
1. In the top menu, click **Runtime > Run all** (or run each cell sequentially).
2. When prompted in Cell 2, upload your `MindBurst.zip` file (or let it clone your repository).
3. Buildozer will compile the Android APK.
4. The final `.apk` file will automatically download to your device!

### Step 2: Upload Project Archive (`MindBurst.zip`)

```python
import os
from google.colab import files

# Check if MindBurst already exists
if not os.path.exists('mindburst'):
    print("Please upload your MindBurst.zip file:")
    uploaded = files.upload()
    for fn in uploaded.keys():
        if fn.endswith('.zip'):
            !unzip -o "{fn}"
            break

print("\n[OK] Project files ready:")
!ls -la
```

### Step 3: Build Android APK with Buildozer

```python
# Compile the Android APK (takes ~10-15 minutes on first run to setup Android NDK & SDK)
!buildozer -v android debug
```

### Step 4: Download the Compiled APK

```python
import glob
from google.colab import files

apk_files = glob.glob('bin/*.apk')
if apk_files:
    apk_path = apk_files[0]
    print(f"\n[SUCCESS] Found generated APK: {apk_path}")
    print("Starting download to your device...")
    files.download(apk_path)
else:
    print("\n[ERROR] No APK found in bin/. Please check the Buildozer log above.")
```
